```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

results_df = pd.read_csv('follicle_detection_figs/model_eval/model_predictions.csv')

results_df


```

```python
fn = results_df.query('predicted_label == 0 and true_label == 1')

for i, row in fn.iterrows():
    image = Image.open(row['image_path'])
    model_output = np.load(row['model_output_path'])
    prediction_mask = np.load(row['mask_path'])
    fig, ax = plt.subplots(1, 3)
    ax[0].imshow(image)
    ax[0].axis('off')
    ax[1].imshow(model_output)
    ax[1].axis('off')
    ax[2].imshow(prediction_mask)
    ax[2].axis('off')
    plt.savefig(f'follicle_detection_figs/model_eval/false_negatives/image_{i}')
    # plt.show()

    
```

```python
tp = results_df.query('predicted_label == 1 and true_label == 1')

for _, row in tp.iterrows():
    image = Image.open(row['image_path'])
    model_output = np.load(row['model_output_path'])
    prediction_mask = np.load(row['mask_path'])
    fig, ax = plt.subplots(1, 3)
    ax[0].imshow(image)
    ax[0].axis('off')
    ax[1].imshow(model_output)
    ax[1].axis('off')
    ax[2].imshow(prediction_mask)
    ax[2].axis('off')
    plt.show()
```
